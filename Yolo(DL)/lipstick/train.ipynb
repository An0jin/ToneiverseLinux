```python
from ultralytics import YOLO
import koreanize_matplotlib
```

```python
model=YOLO('yolo26n.pt')
model.train(
    data='./dataset/data.yaml',
    epochs=300,
    imgsz=640,
    batch=32,
    val=True,
    # --- 색상 관련: 기존 유지 ---
    hsv_h=0.015,
    hsv_s=0.7,
    hsv_v=0.4,
    # --- 구도 관련: 대폭 강화 ---
    degrees=15.0,      # 회전각을 조금 더 확보
    translate=0.3,     # 0.1 -> 0.3으로 강화 (객체를 구석으로 강제 이동)
    scale=0.9,         # 0.5 -> 0.9로 강화 (범위: 0.1 ~ 1.9배. 아주 작은 객체 생성 유도)
    perspective=0.001, # 미세한 원근감 추가 (다양한 각도 대응)
    fliplr=0.5,
    flipud=0.2,        # 립스틱은 뒤집혀 있어도 인식해야 하므로 추가
    mosaic=1.0,        # 필수 유지
    mixup=0.1,
    copy_paste=0.1,    # 객체를 복사해서 다른 배경에 붙임 (중앙 집중 해소에 효과적)
    patience=50,
)
```

```python
model=YOLO('yolo12n.pt')
model.train(
    data='./dataset/data.yaml',
    epochs=300,
    imgsz=640,
    batch=32,
    # --- 색상 정밀도 유지 (보수적 접근) ---
    hsv_h=0.015,       # 아주 최소한의 색조 변화 허용
    hsv_s=0.1,         # 채도는 조명 환경에 따라 변하므로 0.05보다는 약간 상향
    hsv_v=0.2,         # 명암 차이는 실제 환경에서 빈번하므로 조금 더 허용
    
    # --- 기하학적 변형 (최소한의 물리적 법칙 반영) ---
    degrees=180.0,     # 상위 뷰라면 립스틱이 어느 방향으로든 놓일 수 있으므로 필수
    translate=0.1,     # 위치가 조금만 틀어져도 못 잡는 사태 방지
    scale=0.1,         # 미세한 거리 변화 대응
    fliplr=0.5,        
    flipud=0.5,        # 상위 뷰(Top View)이므로 상하 반전도 반드시 포함
    
    # --- 데이터 품질 유지 ---
    mosaic=1.0,        # 작은 객체(립스틱) 탐지에는 켜는 것이 유리 (YOLO의 핵심)
    mixup=0.0,         # 립스틱 색상이 섞이면 안 되므로 끄는 것 유지
    overlap_mask=True, 
    
    patience=50,
    save=True,
)
```
