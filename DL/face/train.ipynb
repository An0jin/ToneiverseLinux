```python
from ultralytics import YOLO
import koreanize_matplotlib
from wakepy import keep
```

```python
# YOLO 사전학습 모델 가중치 로드 (yolo26n.pt)
model = YOLO('yolo26n.pt')
with keep.presenting():
    # 얼굴 톤 탐지 모델 학습 실행
    model.train(
        data='./dataset/data.yaml', # 데이터셋 경로 및 클래스 정보가 담긴 yaml 파일
        epochs=300,                 # 최대 학습 에폭 수
        imgsz=640,                  # 입력 이미지 해상도 (640x640)
        batch=32,                   # 배치 사이즈
        val=True,                   # 학습 중 검증(Validation) 수행 여부
        # --- 색상 증강 설정: 퍼스널컬러 고유 색조 유지를 위해 보수적으로 설정 ---
        hsv_h=0.0,                  # 색조(Hue) 변화 미적용 (얼굴 톤 왜곡 방지)
        hsv_s=0.0,                  # 채도(Saturation) 변화 미적용
        hsv_v=0.1,                  # 명도(Value/밝기) 미세 조절만 허용
        # --- 기하학 증강 설정 ---
        degrees=0.0,                # 회전 변형 없음
        translate=0.1,              # 위치 이동(Translation) 변형 허용
        scale=0.5,                  # 크기 배율 조절
        fliplr=0.5,                 # 좌우 반전(Horizontal Flip) 50% 확률 적용
        mosaic=0.0,                 # 모자이크 증강 비활성화 (얼굴 색조 혼합 방지)
        flipud=0.0,                 # 상하 반전 비활성화 (얼굴 방향 유지)
        mixup=0.0,                  # MixUp 비활성화
        patience=50,                # 50 에폭 동안 성능 개선 없을 시 조기 종료(Early Stopping)
    )
```
